```julia
from matplotlib.pyplot import savefig
from notes import equal

using DifferentialEquations
using Plots
```

```julia
function mutualism_1!(du, u, p, е)
    X, Y = u
    c1, c2, p1, p2 = p

    du[1] = -c1*X + p1*X*Y
    du[2] = -c2*Y + p2*X*Y
end
```

```julia
plt_1_0 = plot(xlabel="Вид Х\n\n",
               ylabel="Вид У",
               label="",
               title="",
               xlims = (0, 6),
               ylims = (0, 6),
               legend=:right,
               aspect_ratio = :equal)

X_0 = 2.5
Y_0 = 4
time = (0.0, 3.5)
params = [0.6, 0.6, 0.2, 0.2]

prob = ODEProblem(mutualism_1!, [X_0, Y_0], time, params)
sol = solve(prob, Tsit5(), reltol=1e-8, abstol=1e-8)

plot!(plt_1_0, [u[1] for u in sol.u], [u[2] for u in sol.u],
      linewidth=2,
      label="а",
      legend=:right, arrow=:head)


X_0 = 2
Y_0 = 4
time = (0.0, 10)
params = [0.6, 0.6, 0.2, 0.2]

prob = ODEProblem(mutualism_1!, [X_0, Y_0], time, params)
sol = solve(prob, Tsit5(), reltol=1e-8, abstol=1e-8)

plot!(plt_1_0, [u[1] for u in sol.u], [u[2] for u in sol.u],
      linewidth=2,
      label="б",
      legend=:right, arrow=:head)

display(plt_1_0)
savefig(plt_1_0, "phases/phase_1_0.png")
```

```julia
plt_1_1 = plot(xlabel="Вид Х\n\n",
               ylabel="Вид У",
               label="",
               title="",
               xlims = (0, 6),
               ylims = (0, 6),
               legend=:right,
               aspect_ratio = :equal)

time = (0.0, 10.0)
params = [0.6, 0.6, 0.2, 0.2]

for X_0 in 0:0.1:4
    Y_0 = 4
    time = (0.0, 5.0)

    prob = ODEProblem(mutualism_1!, [X_0, Y_0], time, params)
    sol = solve(prob, Tsit5(), reltol=1e-8, abstol=1e-8)

    plot!(plt_1_1, [u[1] for u in sol.u], [u[2] for u in sol.u],
          linewidth=0.5,
          label="",
          color=:blue,
          legend=:right)
end

for Y_0 in 0:0.1:4
    X_0 = 4

    prob = ODEProblem(mutualism_1!, [X_0, Y_0], time, params)
    sol = solve(prob, Tsit5(), reltol=1e-8, abstol=1e-8)

    plot!(plt_1_1, [u[1] for u in sol.u], [u[2] for u in sol.u],
          linewidth=0.5,
          label="",
          color=:blue,
          legend=:right)
end
display(plt_1_1)
```

```julia
function mutualism_2!(du, u, p, t)
    X, Y = u
    c1, c2, p1, p2, d1, d2 = p

    du[1] = -c1*X + p1*X*Y/(1+d1*Y)
    du[2] = -c2*Y + p2*X*Y/(1+d2*X)
end
```

```julia
 plt_2_1 = plot(xlabel="Вид Х\n\n",
                ylabel="Вид У",
                label="",
                title="а)",
                xlims = (0, 20),
                ylims = (0, 20),
                legend=:right,
                aspect_ratio = :equal)

for X_0 in 0:20
    for Y_0 in 0:20
        time = (0.0, 50.0)

        prob = ODEProblem(mutualism_2!, [X_0, Y_0], time, [0.6, 0.6, 0.2, 0.2, 0.2, 0.2])
        sol = solve(prob, Tsit5(), reltol=1e-8, abstol=1e-8)

        plot!(plt_2_1, [u[1] for u in sol.u], [u[2] for u in sol.u],
              linewidth=0.5,
              label="",
              color=:blue,
              legend=:right)
    end
end
```

```julia
plt_2_2 = plot(xlabel="Вид Х\n\n",
               ylabel="Вид У",
               label="",
               title="б)",
               xlims = (0, 20),
               ylims = (0, 20),
               legend=:right,
               aspect_ratio = :equal)

for X_0 in 0:0.2:20
    Y_0 = 20
    time = (0.0, 50.0)

    prob = ODEProblem(mutualism_2!, [X_0, Y_0], time, [0.6, 0.6, 0.2, 0.2, 0.2, 0.2])
    sol = solve(prob, Tsit5(), reltol=1e-8, abstol=1e-8)

    plot!(plt_2_2, [u[1] for u in sol.u], [u[2] for u in sol.u],
          linewidth=0.5,
          label="",
          color=:blue,
          legend=:right)
end

for Y_0 in 0:0.2:20
    X_0 = 20
    time = (0.0, 50.0)

    prob = ODEProblem(mutualism_2!, [X_0, Y_0], time, [0.6, 0.6, 0.2, 0.2, 0.2, 0.2])
    sol = solve(prob, Tsit5(), reltol=1e-8, abstol=1e-8)

    plot!(plt_2_2, [u[1] for u in sol.u], [u[2] for u in sol.u],
          linewidth=0.5,
          label="",
          color=:blue,
          legend=:right)
end
```

```julia
plot_2 = plot(plt_2_1, plt_2_2,
                     layout=@layout([a b]),
                     size=(1000, 500),
                     plot_title="Фазовые портреты с насыщением")
display(plot_2)
savefig(plot_2, "phases/phase_2.png")
```

```julia
function mutualism_3!(du, u, p, t)
    X, Y = u
    c1, c2, p1, p2, d1, d2, e1, e2 = p

    du[1] = -c1*X + p1*X*Y/(1+d1*Y) - e1*X*X
    du[2] = -c2*Y + p2*X*Y/(1+d2*X) - e2*Y*Y
end
```

```julia
plt_3_1 = plot(xlabel="Вид Х\n\n",
               ylabel="Вид У",
               label="",
               title="а)",
               xlims = (0, 20),
               ylims = (0, 20),
               legend=:right,
               aspect_ratio=:equal)

for X_0 in 0:20
    for Y_0 in 0:20
        time = (0.0, 50.0)

        prob = ODEProblem(mutualism_3!, [X_0, Y_0], time, [0.6, 0.6, 0.2, 0.2, 0.2, 0.2, 0.01, 0.01])
        sol = solve(prob, Tsit5(), reltol=1e-8, abstol=1e-8)

        plot!(plt_3_1, [u[1] for u in sol.u], [u[2] for u in sol.u],
              linewidth=0.5,
              label="",
              color=:blue,
              legend=:right)
    end
end
```

```julia
plt_3_2 = plot(xlabel="Вид Х\n\n",
               ylabel="Вид У",
               label="",
               title="",
               xlims = (0, 20),
               ylims = (0, 20),
               legend=:right,
               aspect_ratio=:equal)

params = [0.6, 0.6, 0.2, 0.2, 0.2, 0.2, 0.01, 0.01]
time = (0.0, 50.0)

for X_0 in 0:0.4:20
    Y_0 = 20

    prob = ODEProblem(mutualism_3!, [X_0, Y_0], time, params)
    sol = solve(prob, Tsit5(), reltol=1e-8, abstol=1e-8)

    plot!(plt_3_2, [u[1] for u in sol.u], [u[2] for u in sol.u],
          linewidth=0.5,
          label="",
          color=:blue,
          legend=:right)
end

for Y_0 in 0:0.4:20
    X_0 = 20

    prob = ODEProblem(mutualism_3!, [X_0, Y_0], time, params)
    sol = solve(prob, Tsit5(), reltol=1e-8, abstol=1e-8)

    plot!(plt_3_2, [u[1] for u in sol.u], [u[2] for u in sol.u],
          linewidth=0.5,
          label="",
          color=:blue,
          legend=:right)
end
```
